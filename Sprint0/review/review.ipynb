```
public class Plant
{
    public string Name;
    public string Type;
    public float Height;

    public Plant(string name, string type, float height)
    {
        Name = name;
        Type = type;
        Height = height;
    }

    public virtual void Bloom()
    {
        Console.WriteLine($"Растение {Name} цветёт.");
    }

    public void Grow(float growth)
    {
        Height= Height +growth;
        Console.WriteLine($"Растение {Name} выросло на {growth} см. Текущий рост: {Height}");
    }
}

public class Flower : Plant
{
    public string Color;
    public int Leafcount;

    public Flower(string name, float height, string color, int leafcount) 
        : base(name, "Цветок", height) 
    {
        Color=color;
        Leafcount=leafcount;
    }

    public override void Bloom()
    {
        Console.WriteLine($"Цветок {Name} распускается и радует глаз!");
    }

    public void wither()
    {
        Console.WriteLine($"Цветок {Name} начинает увядать.");
    }
}

public class Tree : Plant
{
    public int Age;
    public string Leaftype;

    public Tree(string name, float height, int age, string leaftype)
        : base(name, "Дерево", height)
    {
        Age=age;
        Leaftype=leaftype;
    }

    public override void Bloom()
    {
        Console.WriteLine($"Дерево {Name} покрывается новыми листьями!");
    }
    public void Agegrowth(int years)
    {
        Age= +years;
        Console.WriteLine($"Дерево {Name} выросло на {years} лет. Настоящий возраст: {Age} лет.");
    }
}

Flower rose = new Flower("Роза", 45.0f, "Красный", 15);
Tree dub = new Tree("Дуб", 300.0f, 10, "Лиственные");

Console.WriteLine("Рост и увядание цветка:");
rose.Grow(5.5f);
rose.wither();
Console.WriteLine("Рост и старение дерева:");
dub.Grow(20.0f);
dub.Agegrowth(2);

Plant[] plants = { rose, dub};

foreach (Plant plant  in plants)
{
    plant.Bloom();
}
```
